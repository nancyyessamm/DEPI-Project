What code would you use to set up a view of all movies that did not get released?


```python
%%sql
CREATE VIEW unreleased_movies AS
SELECT *
FROM movies
WHERE release_status = 'Unreleased';
```

### Question 7
What is the most popular movie that was made after 01/01/2000 with a budget of more than $100 000 000? (Hint: Use the popularity field in the Movies table. Larger numbers are more popular.)


```python
%%sql
SELECT title, release_date, popularity, budget
FROM movies
WHERE release_date > '2000-01-01' AND budget > 100000000
ORDER BY popularity DESC
LIMIT 1;
```

### Question 12
You have just watched The Flintstones movie and did not find it very funny. What code would delete the entry that links The Flintstones to the Comedy genre?


```python
%%sql
DELETE FROM genremap
WHERE movie_id = (SELECT movie_id FROM movies WHERE title = 'The Flintstones')
AND genre_id = (SELECT genre_id FROM genres WHERE genre_name = 'Comedy');
```

2. Analyze Movie Budgets and Revenues

```python
%%sql
SELECT title, budget, revenue
FROM movies
WHERE budget > 0 AND revenue > 0;
```

3. Examine the Distribution of Movies by Language

```python
%%sql
SELECT original_language, COUNT(*) AS movie_count
FROM movies
GROUP BY original_language;
```

4. Retrieve Movies Based on Their Release Year

```python
%%sql
SELECT title, release_date
FROM movies
WHERE strftime('%Y', release_date) = '2020';
```

5. Analyze Different Genres

```python
%%sql
SELECT g.genre_name, COUNT(gm.movie_id) AS movie_count
FROM genres g
JOIN genremap gm ON g.genre_id = gm.genre_id
JOIN movies m ON gm.movie_id = m.movie_id
GROUP BY g.genre_name;
```

6. Identify the Top Production Companies by Number of Movies

```python
%%sql
SELECT pc.production_company_name AS production_company,
       COUNT(pm.movie_id) AS movie_count
FROM productioncompanies pc
JOIN productioncompanymap pm ON pc.production_company_id = pm.production_company_id
GROUP BY pc.production_company_name
ORDER BY movie_count DESC
LIMIT 10;
```

7. Visualizing the Data

```python
%%sql
CREATE VIEW genre_analysiss AS
SELECT
    g.genre_name,
    COUNT(m.movie_id) AS movie_count
FROM
    genres g
JOIN
    genremap gm ON g.genre_id = gm.genre_id
JOIN
    movies m ON gm.movie_id = m.movie_id
GROUP BY
    g.genre_name;
```

Query the View

```python
%%sql
SELECT * FROM genre_analysiss;

```

8. Closing the Analysis

```python
%%sql
DROP VIEW IF EXISTS genre_analysiss;
```
